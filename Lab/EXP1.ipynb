```python
import pandas as pd # Training data
data = [
['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]
# Column names
columns = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport']
# Create DataFrame
df = pd.DataFrame(data, columns=columns)
print("Training Data:\n")
print(df)

# Initialize hypothesis
hypothesis = ['0'] * (len(columns) - 1)
print("\nInitial Hypothesis:")
print(hypothesis)

# Apply FIND-S Algorithm
for i in range(len(data)):
    # Consider only positive examples
    if data[i][-1] == 'Yes':
        for j in range(len(hypothesis)):
            # If hypothesis is 0, replace with attribute value
            if hypothesis[j] == '0':
                hypothesis[j] = data[i][j]
            # If attribute values differ, generalize with '?'
            elif hypothesis[j] != data[i][j]:
                hypothesis[j] = '?'
        print("\nHypothesis after positive example", i+1)
        print(hypothesis)

# Final hypothesis
print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```
