```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
# Sample Dataset
data = {
    'Income': [25000, 50000, 75000, 30000, 60000, 80000, 40000, 90000],
    'Age': [25, 35, 45, 23, 40, 50, 30, 55],
    'Loan_Amount': [10000, 20000, 30000, 12000, 25000, 35000, 15000, 40000],
    'Credit_Score': ['Low', 'Medium', 'High', 'Low', 'Medium', 'High', 'Medium', 'High']
}
df = pd.DataFrame(data)
# Encode Labels
df['Credit_Score'] = df['Credit_Score'].map({'Low': 0, 'Medium': 1, 'High': 2})
# Features and Target
X = df[['Income', 'Age', 'Loan_Amount']]
y = df['Credit_Score']
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
# Model (Random Forest)
model = RandomForestClassifier()
model.fit(X_train, y_train)
# Prediction
predictions = model.predict(X_test)
# Output
print("Test Data:\n", X_test)
print("\nPredicted Credit Scores (0=Low,1=Medium,2=High):\n", predictions)
```
